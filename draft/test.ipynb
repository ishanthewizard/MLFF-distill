```python
from rdkit import Chem
from rdkit.Chem import Draw

sdf_file = "/global/homes/y/yuejian/project/MLFF-distill/yuejian/ligandboundconf3/xtb_local_min/0A9_3QGO_A_407/0A9_3QGO_A_407.sdf"
# RDKit's SDMolSupplier expects a file-like object, so read directly from file
suppl = Chem.SDMolSupplier(sdf_file, removeHs=False)
mols = [mol for mol in suppl if mol is not None]
if mols:
    # Draw the first molecule (can change index if you want a different one)
    img = Draw.MolToImage(mols[0], size=(400, 400), kekulize=True)
    display(img)
else:
    print("No molecules could be loaded from the SDF file.")
```
