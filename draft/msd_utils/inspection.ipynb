```python
# Example: Load ATOMIC trajectory and print info from user-provided .traj file

from ase.io import Trajectory

traj_path = "/global/homes/y/yuejian/project/MLFF-distill/m4558/nitesh_original_trajs/first_100ps_traj_to_send/md_omol_natfsi_dme_s1p1.traj"

with Trajectory(traj_path) as traj:
    print(f"Trajectory contains {len(traj)} frames.")
    frame0 = traj[0]
    print("First frame:")
    import shutil

    symbols = frame0.get_chemical_symbols()
    term_width = shutil.get_terminal_size((80, 20)).columns

    # Format symbols for printing based on terminal width
    import textwrap
    symbols_str = ", ".join(symbols)
    wrapped = textwrap.fill(symbols_str, width=term_width - 16)  # indent + label length

    print("\tSymbols:")
    for line in wrapped.splitlines():
        print("\t  " + line)

```
