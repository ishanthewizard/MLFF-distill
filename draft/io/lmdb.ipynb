```python
import sys
sys.path.append("../..")
import torch
from src_v2.distill_datasets import LmdbDataset
import matplotlib.pyplot as plt
import lmdb
import pickle
from tqdm import tqdm
```

```python
src = "/home/yuejian/project/MLFF-distill/OMOL/4M/subsets/OMol_subset_label_finite_difference/protein_ligand_pockets_train_4M/final_labels/train_forces"
target = "/home/yuejian/project/MLFF-distill/OMOL/TOY/labels/ligand_pocket_1000_esen_md_fd_dt_0_0001/train_forces/data.lmdb"
```

```python
fd_dataset = LmdbDataset(
    folder_path=src,
)
```

```python
env = lmdb.open(target, map_size=1099511627776 * 2)  # ~1 GB


for i,data in tqdm(enumerate(fd_dataset),total=len(fd_dataset)):
    if i >= 1000:
        break
    with env.begin(write=True) as txn:
        idx = str(int(i))
        txn.put(idx.encode(), data.numpy().tobytes())

env.close()
```
