```python
import pandas as pd
from ase import Atoms
from rdkit import Chem
```

```python
parquet_path = "/home/yuejian/project/MLFF-distill/data/ligandboundconf3.0corrected_500/subset_500_global_500_local.parquet"
df = pd.read_parquet(parquet_path)
print(df.head())
```

```python
# get a mol byte from the parquet file and transform it to rdkit mol object
mol_bytes = df.iloc[0]['xtb_local_min_mol']
mol = Chem.Mol(mol_bytes)
# Get atom numbers (atomic numbers)
atom_numbers = [atom.GetAtomicNum() for atom in mol.GetAtoms()]

# Get atom coordinates
conf = mol.GetConformer()
positions = conf.GetPositions()

# Get charge and spin (multiplicity)
# Try to get from DataFrame if available, else default to 0 and 1
charge = df.iloc[0]['charge'] if 'charge' in df.columns else Chem.GetFormalCharge(mol)
spin = df.iloc[0]['spin'] if 'spin' in df.columns else Chem.GetSpinMultiplicity(mol) if hasattr(Chem, "GetSpinMultiplicity") else 1


print("Charge:", charge)
print("Spin multiplicity:", spin)
print("Atom numbers:", atom_numbers)
print("Positions:", positions)
print("Molecule:", mol)

```
