```python
import os
import random
from ase.io import read, write

file_path = '/data/ishan-amin/OMOL/electrolytes_application/trajs/electrolytes_augmented_1000.xyz'

# Read all structures from the xyz file
atoms_list = read(file_path, index=':')

# Shuffle and split into 90% train, 10% val
num_structures = len(atoms_list)
indices = list(range(num_structures))
random.shuffle(indices)

split_idx = int(0.9 * num_structures)
train_indices = indices[:split_idx]
val_indices = indices[split_idx:]

train_structures = [atoms_list[i] for i in train_indices]
val_structures = [atoms_list[i] for i in val_indices]

# Create output folders
os.makedirs('train', exist_ok=True)
os.makedirs('val', exist_ok=True)

# Write to xyz files
write('train/train.xyz', train_structures)
write('val/val.xyz', val_structures)
```

```python
# Extract energies, forces, and stress from the 0th structure
atoms0 = atoms_list[0]
energy0 = atoms0.get_potential_energy()
forces0 = atoms0.get_forces()
stress0 = atoms0.get_stress()
print(atoms0.calc.results)

print("Energy (0th structure):", energy0)
print("Forces (0th structure):", forces0)
print("Stress (0th structure):", stress0)
```
