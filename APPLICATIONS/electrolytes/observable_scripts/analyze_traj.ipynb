```python
from ase.io import read
import matplotlib.pyplot as plt
import numpy as np
```

```python
traj_path = "/global/homes/y/yuejian/project/MLFF-distill/yuejian/MD_data_rest/md_omol_natfsi_dme_s1p1/md_omol_natfsi_dme_s1p1.traj"
images = read(traj_path, index=":")

print(f"Loaded {len(images)} frames from {traj_path}")

# Prepare arrays for each property
potential_energies = [img.get_potential_energy() for img in images]
kinetic_energies = [img.get_kinetic_energy() for img in images]
volumes = [img.get_volume() for img in images]
temperatures = [img.get_temperature() for img in images]

timesteps = np.arange(len(images))

fig, axs = plt.subplots(4, 1, figsize=(10, 16), sharex=True)

axs[0].plot(timesteps, potential_energies, label='Potential Energy')
axs[0].set_ylabel('Potential Energy (eV)')
axs[0].legend()

axs[1].plot(timesteps, kinetic_energies, label='Kinetic Energy', color='orange')
axs[1].set_ylabel('Kinetic Energy (eV)')
axs[1].legend()

axs[2].plot(timesteps, volumes, label='Volume', color='green')
axs[2].set_ylabel('Volume ($\mathrm{\AA}^3$)')
axs[2].legend()

axs[3].plot(timesteps, temperatures, label='Temperature', color='red')
axs[3].set_ylabel('Temperature (K)')
axs[3].set_xlabel('Frame')
axs[3].legend()

plt.tight_layout()
```
